```python
%%writefile desplazamientos.c
#include <stdio.h>

int main() {
    // Usamos unsigned char (1 byte = 8 bits) para trabajar de forma limpia
    unsigned char estado = 40; // Binario: 0010 1000
    int pos = 3;               // Bit objetivo (cuarto bit desde la derecha)

    printf("==================================================\n");
    printf("ESTADO INICIAL: %d (Binario: 0010 1000)\n", estado);
    printf("==================================================\n\n");

    // ------------------------------------------------------------------
    // PARTE 1: DESPLAZAMIENTO A LA IZQUIERDA (<<) -> MODIFICAR BITS
    // ------------------------------------------------------------------
    printf("--- 1. MODIFICACIONES CON IZQUIERDA (<<) ---\n");

    // Crear la máscara trasladando el 1 a la posición 'pos'
    unsigned char mascara = (1 << pos); // Binario: 0000 1000
    printf("Mascara (1 << %d): %d (Binario: 0000 1000)\n", pos, mascara);

    // A. Encender bit (OR |)
    unsigned char estado_encendido = estado | (1 << pos);
    printf("Encender bit %d (|): %d\n", pos, estado_encendido);

    // B. Apagar bit (AND con NOT & ~)
    unsigned char estado_apagado = estado & ~(1 << pos);
    printf("Apagar bit %d (& ~): %d\n", pos, estado_apagado);

    // C. Alternar / Invertir bit (XOR ^)
    unsigned char estado_alternado = estado ^ (1 << pos);
    printf("Alternar bit %d (^): %d\n\n", pos, estado_alternado);

    // ------------------------------------------------------------------
    // PARTE 2: DESPLAZAMIENTO A LA DERECHA (>>) -> EXTRAER Y LEER BITS
    // ------------------------------------------------------------------
    printf("--- 2. LECTURA Y EXTRACCIÓN CON DERECHA (>>) ---\n");

    // A. Extraer el valor directo (0 o 1) traendolo a la posición 0
    int bit_extraido = (estado >> pos) & 1;
    printf("Valor del bit en la pos %d: %d\n", pos, bit_extraido);

    // B. Probar con otra posición que valga 0 (por ejemplo, pos = 1)
    int pos_cero = 1;
    int bit_cero = (estado >> pos_cero) & 1;
    printf("Valor del bit en la pos %d: %d\n", pos_cero, bit_cero);

    // C. Uso matemático: División rápida entera entre 2^n
    unsigned char division_por_2 = estado >> 1; // 40 / 2^1 = 20
    unsigned char division_por_4 = estado >> 2; // 40 / 2^2 = 10
    printf("Division rapida (%d >> 1 = %d / 2): %d\n", estado, estado, division_por_2);
    printf("Division rapida (%d >> 2 = %d / 4): %d\n", estado, estado, division_por_4);

    return 0;
}
```

int ancho = 5;

Calcula:

$$ Área = largo \times ancho $$

Y muestra:

```python
%%writefile ejercicio3.c

#include <stdio.h>

int largo = 10;
int ancho = 5;

int main(){

  int area = largo * ancho;

  printf("el largo es %d\n", largo);

  printf("el ancho es %d\n", ancho);

  printf("el area del rectangulo es %d\n", area);

  return 0;
}
```
